```python
class SharedDict:
    """A mutable shared dictionary container."""
    def __init__(self, data=None):
        self.data = dict(data) if data else {}

    def __getitem__(self, key):
        return self.data[key]

    def __setitem__(self, key, value):
        self.data[key] = value

    def __iter__(self):
        return iter(self.data)

    def items(self):
        return self.data.items()

    def keys(self):
        return self.data.keys()

    def values(self):
        return self.data.values()

    def update(self, other):
        self.data.update(other)

    def __len__(self):
        return len(self.data)

    def __contains__(self, key):
        return key in self.data

    def __repr__(self):
        return f"<SharedDict {self.data}>"

class Tree:
    """A mutable shared dictionary container."""
    def __init__(self, s1=None):
        self.parameter = SharedDict({s1:s1})
s1 = Tree('s1')
s2 = Tree('s2')
s3 = Tree('s3')
```

```python
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```

```python
s2.parameter = s1.parameter
s3.parameter = s2.parameter
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```

```python
s1.parameter['s1'] = 4
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```

```python
s3.parameter['s1'] = 4
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```

```python
a = {k:v for k,v in s1.parameter.items()}
print(a)

a.update({'kk':3})
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)

s2.parameter.update(a)
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```

```python
s3.parameter.update({'s1':'x'})
print(s1.parameter)
print(s2.parameter)
print(s3.parameter)
```
